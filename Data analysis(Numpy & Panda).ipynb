# Numpy Array Operations

```python
import numpy as np
```

```python
# Create a 1D NumPy Array
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
```

```python
temperatures_w1
```

```python
# Inspection and Properties
print("Shape:", temperatures_w1.shape)
```

```python
print("Data Type:", temperatures_w1.dtype)
```

```python
print("Number of Elements:", temperatures_w1.size)
```

```python
# Array Operations
fahrenheit = (temperatures_w1 * 9/5) + 32
print("Fahrenheit:", fahrenheit)
```

```python
print("Maximum:", np.max(temperatures_w1))
```

```python
print("Minimum:", np.min(temperatures_w1))
```

```python
print("Mean:", np.mean(temperatures_w1))
```

```python
# Array Slicing and Indexing
# The temperatures for the first three days of the week
print("First 3 Days:", temperatures_w1[:3])
```

```python
# The temperatures for the weekend (last two days)
print("Weekend:", temperatures_w1[-2:])
```

```python
# The temperatures for the middle three days of the week 
print("Middle 3 Days:", temperatures_w1[2:5])
```

```python
# Create a 2D Array
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])
```

```python
temperatures
```

```python
# Inspect and Slice the 2D Array 
print("Shape:", temperatures.shape)
```

```python
print("Data Type:", temperatures.dtype)
```

```python
print("Total Elements:", temperatures.size)
```

```python
#  Extract  temperature of each week
print("Week 1:", temperatures[0])
```

```python
print("Week 2:", temperatures[1])
```

```python
# The weekends 
print("Weekends:", temperatures[:, -2:])
```

# Pandas Series

```python
import pandas as pd
```

```python
# Create a Pandas Series
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)
```

```python
print(marks)
```

```python
# Indexing and Slicing
#Access 1st rank using integer position 
print("1st Rank:", marks.iloc[0])
```

```python
# Retrieve top 3 ranks using loc
print(marks.iloc[0])
```

```python
# Retrieve top 3 ranks using loc
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
```

```python
# Retrieve 3rd rank using iloc
print("3rd Rank:", marks.iloc[2])
```

```python
# Boolean filtering
print("Marks Greater Than 90:")
print(marks[marks > 90])
```

```python
# Manipulating Series
# Modify 1st rank mark
marks.loc['Rank1'] = 100
print(marks)
```

```python
# Remove last rank
marks = marks.drop('Rank5')
print(marks)
```

```python
# Calculate CGPA
cgpa = marks / 10
print("CGPA:")
print(cgpa)
```

# Pandas DataFrame

```python
# Creating Pandas DataFrame
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': ['Electronics', 'Clothing', 'Electronics',
                        'Furniture', 'Clothing', 'Electronics',
                        'Furniture', 'Clothing', 'Furniture', 'Electronics'],

    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)
```

```python
#  Data Exploration
# Display DataFrame
print(transactions)
```

```python
# Display first 5 rows
print(transactions.head())
```

```python
# Display last 5 rows
print(transactions.tail())
```

```python
# Shape
print(transactions.shape)
```

```python
# Column names
print(transactions.columns)
```

```python
# Data types
print(transactions.dtypes)
```

```python
# Display selected columns
print(transactions[['ProductCategory', 'Amount']])
```

```python
# Retrieve last 3 columns
print(transactions.iloc[:, -3:])
```

```python
# Filter North region and Amount greater than 200
print(transactions[(transactions['Region'] == 'North') &
                   (transactions['Amount'] > 200)])
```

```python
# Value counts
print(transactions['ProductCategory'].value_counts())
```

```python
# Unique regions
print(transactions['Region'].unique())
```

```python
# Mean amount by region
print(transactions.groupby('Region')['Amount'].mean())
```

```python
# 3. DataFrame Manipulation

# Modify Amount for TransactionID 102
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
```

```python
# Add Discount column
transactions['Discount'] = transactions['Amount'] * 0.10
print(transactions)
```

```python
# Remove TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]
print(transactions)
```

```python
# Delete Discount column
transactions = transactions.drop('Discount', axis=1)
print(transactions)
```
